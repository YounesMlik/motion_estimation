```python
def hstack_videos(*videos: Iterable[BGRImage]) -> Iterable[BGRImage]:
    return map(lambda *vs: BGRImage(np.hstack((*vs,))), *videos)
```

```python
video = read_video_from_jpeg_frames(video_paths[4])


show_video(
    hstack_videos(
        video,
        ViBe.process_video(
            video, ViBeConfig(nb_samples=20, subsampling_factor=2, radius=30)
        ),
        FrameDifference.process_video(video, 20),
    )
)
```
